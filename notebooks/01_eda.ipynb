CICIDS2017 & CICIoT2023

```python
import sys; sys.path.insert(0, '..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load processed data
X_train = np.load('../data/processed/cicids2017/X_train.npy')
yb_train = np.load('../data/processed/cicids2017/yb_train.npy')
print('Train shape:', X_train.shape)
print('Class balance:', dict(zip(*np.unique(yb_train, return_counts=True))))
```

```python
# PCA component distribution
fig, axes = plt.subplots(2, 4, figsize=(16, 6))
for i, ax in enumerate(axes.flat):
    ax.hist(X_train[:, i], bins=50, color='steelblue', alpha=0.7)
    ax.set_title(f'PC{i+1}')
plt.suptitle('PCA Component Distributions – CICIDS2017 (Train)')
plt.tight_layout()
plt.savefig('../results/eda_pca_dist.png', dpi=150)
plt.show()
```
